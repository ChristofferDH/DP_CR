# Counterfactual simulations

```python
%load_ext autoreload
%autoreload 2
import numpy as np
import matplotlib.pyplot as plt
from Files import Model 
from Files import Estimation
```

### Comparing wealthy and poor households and how they respond to transitory income shocks

```python
model_wealthy = Model.DurableBufferStock()
model_wealthy.modelSetup()
model_wealthy.createGrids()
par = model_wealthy.par
par.counter_factual = 1
model_wealthy.solve()
sol = model_wealthy.sol
```

```python
model_poor = Model.DurableBufferStock()
model_poor.modelSetup()
model_poor.createGrids()
par_poor = model_poor.par
par_poor.counter_factual = 1
model_poor.solve()

sol_poor = model_poor.sol
```

```python
par_poor = model_poor.par
par_poor.counter_factual = 1
model_poor.simulate()
sim_poor = model_poor.sim


par = model_wealthy.par
par.counter_factual = 1
par.sim_d_ini[:] *= 2
par.sim_p_ini[:] *= 2
model_wealthy.simulate()
sim = model_wealthy.sim
```

```python
fig = plt.figure(figsize=(8,5))
ax = fig.add_subplot(1,1,1)
ax.plot(np.arange(par.T), np.mean(sim.c, axis=1), label='Wealthy households')
ax.plot(np.arange(par.T), np.mean(sim_poor.c, axis=1), label='Poor households')
shock_periods = [1, 11, 21, 31, 41]
for t in shock_periods:
    ax.axvline(t, color='red', linestyle='--', alpha=0.4, linewidth=1)

ax.set_xlabel(f"t")
ax.set_ylabel(f"Non-durable consumption $c_t$")
ax.set_title(f'Average non-durable consumption')
ax.legend()
#ax.set_ylim(bottom=0.5,top=1.5)
fig.savefig("average_non_durable_consumption3.pdf",
            format="pdf",
            bbox_inches="tight")

plt.show()
```

```python
fig = plt.figure(figsize=(8,5))
ax = fig.add_subplot(1,1,1)
ax.plot(np.arange(par.T), np.mean(sim.d, axis=1), label='Wealthy households')
ax.plot(np.arange(par.T), np.mean(sim_poor.d, axis=1), label='Poor households')
shock_periods = [1, 11, 21, 31, 41]
for t in shock_periods:
    ax.axvline(t, color='red', linestyle='--', alpha=0.4, linewidth=1)

ax.set_xlabel(f"t")
ax.set_ylabel(f"Durable consumption $d_t$")
ax.set_title(f'Average durable consumption')
ax.legend()
fig.savefig("average_durable_consumption3.pdf",
            format="pdf",
            bbox_inches="tight")

plt.show()
```

```python
print(np.mean(sim.d)/np.mean(sim.c))
print(np.mean(sim_poor.d)/np.mean(sim_poor.c))
```

### Adjustment cost counterfactual

```python
model_ac = Model.DurableBufferStock()
model_ac.modelSetup()
model_ac.createGrids()
par_ac = model_ac.par
par_ac.tau = 0.3
model_ac.solve()
model_ac.simulate()
sim_ac = model_ac.sim
```

```python
fig = plt.figure(figsize=(8,5))
ax = fig.add_subplot(1,1,1)
ax.plot(np.arange(par.T), np.mean(sim.c, axis=1), label=r'$\tau = 0.1$')
ax.plot(np.arange(par_ac.T), np.mean(sim_ac.c, axis=1), label=r'$\tau = 0.3$')

ax.set_xlabel(f"t")
ax.set_ylabel(f"Non-durable consumption, $c_t$")
ax.set_title(f'Average non-durable consumption')
ax.legend()
fig.savefig("tau_c.pdf",format="pdf",bbox_inches="tight")
plt.show()
```

```python
fig = plt.figure(figsize=(8,5))
ax = fig.add_subplot(1,1,1)
ax.plot(np.arange(par.T), np.mean(sim.d, axis=1), label=r'$\tau = 0.1$')
ax.plot(np.arange(par_ac.T), np.mean(sim_ac.d, axis=1), label=r'$\tau = 0.3$')
ax.set_xlabel(f"t")
ax.set_ylabel(f"Durable consumption, $d_t$")
ax.set_title(f'Average durable consumption')
ax.legend()
fig.savefig("ces_d.pdf",
            format="pdf",
            bbox_inches="tight")
fig.savefig("tau_d.pdf",format="pdf",bbox_inches="tight")
plt.show()
```
